```python
import numpy as np
```

```python
# Set up Data
# Distantce to next location, Gas price
data = np.array([
    [100, 3.2],
    [100, 3.5],
    [100, 3.7],
    [100, 3.4],
    [100, 3.6],
    [100, 3.3],
    [100, 3.8],
])

# Parameters
MAX_FUEL = 200
```

# Set up the Dynamic Programming Formulation

We assume that we can only fill up exactly much is needed to get to a place

```python
# Init
best_cost = np.full(len(data) + 1, np.inf) # Best cost to get to the end
best_policy = np.zeros_like(best_cost, dtype=int) # Best policy: next location to go to

# Base Case 
best_cost[-1] = 0

for i in range(len(data) - 1, -1, -1):
    
    # For each location ahead (including the final destination), find the cost of filling up to get there
    for j in range(i + 1, len(data) + 1):
        
        # Calculate distance from location i to location j
        if j < len(data):
            distance = np.sum(data[i:j, 0])
        else:
            # j is the final destination, sum all remaining distances
            distance = np.sum(data[i:, 0])
        
        # Check if it is within range of Max Fuel
        if distance > MAX_FUEL:
            continue

        # Calculate the cost of filling up exactly the amount needed to get to location j
        # CONSTRAINT: We can only fill exactly enough fuel to reach the next location
        # (no more, no less) - this is enforced by buying exactly 'distance' units
        cost = data[i, 1] * distance

        # Update the best cost and policy
        if best_cost[i] > cost + best_cost[j]:
            best_cost[i] = cost + best_cost[j]
            best_policy[i] = j  # Store the next location to go to
```

```python
# Print the optimal path
print(f"Optimal cost: ${best_cost[0]:.2f}")
print("\nOptimal path:")
current = 0
while current < len(data):
    next_loc = best_policy[current]
    if next_loc < len(data):
        distance = np.sum(data[current:next_loc, 0])
        cost_at_current = data[current, 1] * distance
        print(f"Location {current}: Fill {distance:.0f} units at ${data[current, 1]:.2f}/unit = ${cost_at_current:.2f} -> Go to location {next_loc}")
    else:
        distance = np.sum(data[current:, 0])
        cost_at_current = data[current, 1] * distance
        print(f"Location {current}: Fill {distance:.0f} units at ${data[current, 1]:.2f}/unit = ${cost_at_current:.2f} -> Reach destination")
        break
    current = next_loc
```
