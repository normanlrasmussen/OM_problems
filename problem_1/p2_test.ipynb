```python
import numpy as np
```

```python
# Set up Data
# Distantce to next location, Gas price
data = np.array([
    [100, 3.2],
    [100, 3.5],
    [100, 3.7],
    [100, 3.4],
    [100, 3.6],
    [100, 3.3],
    [100, 3.8],
])

# Parameters
MAX_FUEL = 250
```

# Set up the Dynamic Programming Formulation

We assume that we can fill up the gas can as much or s little as we want

```python
# Init
best_cost = np.full((len(data) + 1, MAX_FUEL + 1), np.inf) # Best cost to get to the end, gas level
best_policy = np.zeros_like(best_cost, dtype=int) # Best policy: next location to go to

# Base Case, we want to end with 0 gas
best_cost[-1, 0] = 0

# Reverse iterate over each location
for i in range(len(data) - 1, -1, -1):

    # Iterate over each possible gas level
    for g in range(MAX_FUEL + 1):
        
        # Iterate over every possible amount of gas to fill up to find the best cost
        for g_use in range(MAX_FUEL + 1):

            # Check if filling up g_use gallons would exceed max capacity
            if g + g_use > MAX_FUEL:
                continue

            # Calculate resulting gas level after traveling
            resulting_gas = int(g + g_use - data[i, 0])
            
            # Skip if resulting gas level is invalid
            if resulting_gas < 0 or resulting_gas > MAX_FUEL:
                continue

            # Get the cost of filling up g_use gallons and going to the next location
            cost = data[i, 1] * g_use + best_cost[i + 1, resulting_gas]

            # Update the best cost and policy
            if best_cost[i, g] > cost:
                best_cost[i, g] = cost
                best_policy[i, g] = g_use

```

```python
# Print the optimal path
print(f"Optimal cost: ${best_cost[0, 0]:.2f}")
print("\nOptimal path:")

current_gas = 0  # Start with 0 gas at location 0
for i in range(len(data)):
    g_use = best_policy[i, current_gas]
    print(f"Location {i}: Fill up {g_use} gallons")
    
    # Update gas level: current gas + fill up - distance to next location
    current_gas = int(current_gas + g_use - data[i, 0])
```
